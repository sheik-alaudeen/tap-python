```python
#task3
def read_csv(path):

    data = []

    with open(path, 'r', encoding='utf-8', newline='') as file:

        reader = csv.DictReader(file, delimiter=';')

        for row in reader:
            data.append(dict(row))

    return data
```

```python
#task3
orders_path = INPUT_FOLDER / 'orders.csv'

orders_data = read_csv(orders_path)

print("First two records:")
print(orders_data[:2])

print("Total records:", len(orders_data))
```

```python
#task4
def convert_to_snake_case(column_name):

    column_name = column_name.strip()
    column_name = column_name.lower()
    column_name = column_name.replace(" ", "_")

    return column_name
```

```python
#task4
print(convert_to_snake_case("Customer ID"))
print(convert_to_snake_case("Order Status"))
print(convert_to_snake_case("Product Price"))
print(convert_to_snake_case("Freight Value"))
```

```python
#task5
column_mapping = {
    "Customer ID": "customer_id",
    "Order ID": "order_id",
    "Price": "price",
    "Freight Value": "freight_value",
    "Order Status": "order_status"
}
```

```python
#task5
def map_columns(columns, mapping):

    mapped_columns = []

    for column in columns:

        new_column = mapping.get(column, column)

        mapped_columns.append(new_column)

    return mapped_columns
```

```python
#task5
columns = [
    "Customer ID",
    "Order ID",
    "Price",
    "Freight Value",
    "Order Status"
]

standard_columns = map_columns(columns, column_mapping)

print("Original columns:", columns)
print("Standard columns:", standard_columns)
```

```python
#task6
def convert_to_float(data, columns):

    for row in data:
        for column in columns:
            row[column] = float(row[column])

    return data

def convert_to_datetime(data, columns):

    for row in data:
        for column in columns:
            row[column] = pd.to_datetime(row[column])

    return data
```

```python
items_data = read_csv(INPUT_FOLDER / 'order_items.csv')
orders_data = read_csv(INPUT_FOLDER / 'orders.csv')

items_data = convert_to_float(
    items_data,
    ['price', 'freight_value']
)

date_columns = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]

orders_data = convert_to_datetime(orders_data, date_columns)

print("Price and Freight:")
print(items_data[:2])

print("\nOrder Timestamps:")
print(orders_data[:2])

print("\nData Types:")
print("Price:", type(items_data[0]['price']))
print("Freight:", type(items_data[0]['freight_value']))
print("Timestamp:", type(orders_data[0]['order_purchase_timestamp']))
```

```python
#task7
def calculate_order_totals(data):

    order_totals = {}

    for row in data:

        order_id = row['order_id']

        total = row['price'] + row['freight_value']

        if order_id in order_totals:
            order_totals[order_id] += total
        else:
            order_totals[order_id] = total

    return order_totals
```

```python
order_totals = calculate_order_totals(items_data)

print("Order Totals:")
print(list(order_totals.items()))

print("Total unique orders:", len(order_totals))
```
