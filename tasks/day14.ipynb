```python
#task2
import logging
from datetime import datetime
from pathlib import Path

log_folder = Path("logs")
log_folder.mkdir(exist_ok=True)

timestamp = datetime.now().strftime("%Y-%m-%d_%H-%M-%S")
log_file = log_folder / f"run_{timestamp}.log"

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
    handlers=[
        logging.FileHandler(log_file, encoding="utf-8"),
        logging.StreamHandler()
    ],
    force=True
)

logger = logging.getLogger()
```

```python
clean_data = [
    {"customer_id": "C001", "customer_city": "Chennai", "customer_state": "TN"},
    {"customer_id": "C002", "customer_city": "Coimbatore", "customer_state": "TN"}
]
```

```python
#task4
import csv

import json

with open("clean_data.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=clean_data[0].keys())
    writer.writeheader()
    writer.writerows(clean_data)

with open("clean_data.json", "w", encoding="utf-8") as file:
    json.dump(clean_data, file, indent=4)
```

```python
with open("clean_data.csv", "r", encoding="utf-8") as file:
    print(file.read())

with open("clean_data.json", "r", encoding="utf-8") as file:
    print(file.read())
```

```python
#task5
import time

start_time = time.time()
total_rows = 100
valid_rows = 90
rejected_rows = 10
duration = time.time() - start_time
```

```python
summary = {
    "total_rows": total_rows,
    "valid_rows": valid_rows,
    "rejected_rows": rejected_rows,
    "duration_seconds": round(duration, 2)
}

print(summary)

import json

with open("run_summary.json", "w", encoding="utf-8") as file:
    json.dump(summary, file, indent=4)

with open("run_summary.json", "r", encoding="utf-8") as file:
    print(json.load(file))
```

```python
#task 6
import json

manifest_file = "manifest.json"

with open(manifest_file, "w") as file:
    json.dump([], file)
```

```python
with open(manifest_file, "r") as file:
    processed = json.load(file)

filename = "customers.csv"

if filename in processed:
    print("Already processed. Skipping.")
else:
    print("Processing file...")

    processed.append(filename)

    with open(manifest_file, "w") as file:
        json.dump(processed, file, indent=4)
```

```python
#task10
import json
from pathlib import Path

input_folder = Path("..")
manifest_file = Path("../manifest.json")

if manifest_file.exists():
    with open(manifest_file, "r", encoding="utf-8") as file:
        processed_files = json.load(file)
else:
    processed_files = []

for file_path in input_folder.glob("*.csv"):
    if file_path.name in processed_files:
        print(f"Skipping: {file_path.name}")
    else:
        print(f"Processing: {file_path.name}")
        processed_files.append(file_path.name)

with open(manifest_file, "w", encoding="utf-8") as file:
    json.dump(processed_files, file, indent=4)
```
