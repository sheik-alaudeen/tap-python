```python
#task2
from pathlib import Path
import pandas as pd
data_folder = Path("..")

csv_files = list(data_folder.glob("*.csv"))

dataframes = {}

for file in csv_files:
    dataframes[file.stem] = pd.read_csv(file)
    print(f"Loaded: {file.name} | Rows: {len(dataframes[file.stem])}")

print("\nTotal files loaded:", len(dataframes))
```

```python
#task7
import json
from pathlib import Path

input_folder = Path("..")
manifest_file = Path("../manifest.json")

if manifest_file.exists():
    with open(manifest_file, "r", encoding="utf-8") as file:
        processed_files = json.load(file)
else:
    processed_files = []

for csv_file in input_folder.glob("*.csv"):
    if csv_file.name in processed_files:
        print(f"Skipping already processed: {csv_file.name}")
        logger.info(f"Skipping already processed: {csv_file.name}")
    else:
        print(f"Processing: {csv_file.name}")
        logger.info(f"Processing: {csv_file.name}")

        processed_files.append(csv_file.name)

with open(manifest_file, "w", encoding="utf-8") as file:
    json.dump(processed_files, file, indent=4)

print("Manifest updated!")
```
